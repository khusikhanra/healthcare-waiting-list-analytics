```python
import os
import glob

base_folder = r"D:\HealthCare Data"

print("Folder exists:", os.path.exists(base_folder))
print("Folder:", base_folder)

all_files = []

for pattern in ["**/*.csv", "**/*.xlsx", "**/*.xls"]:
    all_files.extend(
        glob.glob(
            os.path.join(base_folder, pattern),
            recursive=True
        )
    )

print("\nData files found:")
for file in all_files:
    print(file)

print("\nTotal data files:", len(all_files))
```

```python
inpatient_files = sorted(
    glob.glob(os.path.join(base_folder, "InPatient", "*.csv"))
)

outpatient_files = sorted(
    glob.glob(os.path.join(base_folder, "Outpatient", "*.csv"))
)

print("INPATIENT FILES:")
for file in inpatient_files:
    print(file)

print("\nOUTPATIENT FILES:")
for file in outpatient_files:
    print(file)
```

```python
all_files = inpatient_files + outpatient_files

for file in all_files:
    temp_df = pd.read_csv(file)

    print("\n" + "=" * 80)
    print("FILE:", os.path.basename(file))
    print("SHAPE:", temp_df.shape)
    print("COLUMNS:")
    print(temp_df.columns.tolist())
```

```python
dataframes = []

for file in all_files:
    temp_df = pd.read_csv(file)

    # File se year nikalna
    year = os.path.basename(file)[-8:-4]

    # Inpatient / Outpatient identify karna
    if "InPatient" in file:
        patient_type = "Inpatient"
    else:
        patient_type = "Outpatient"

    # Additional information
    temp_df["Year"] = int(year)
    temp_df["Patient_Type"] = patient_type

    dataframes.append(temp_df)

df = pd.concat(dataframes, ignore_index=True)

print("Combined dataset created successfully!")
print("Shape:", df.shape)
```

```python
outlier_summary = []

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    count = ((df[col] < lower) | (df[col] > upper)).sum()
    
    outlier_summary.append({
        "Column": col,
        "Outliers": count
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```
